# **MÓDULO 0: PRIMEROS PASOS CON PYTHON**

Es común usar Excel para analizar datos, pero si el número de datos es grande, Excel se queda corto. **0.1 Variables:** En Excel, guardamos datos en celdas (A1, B2). En Python, usamos variables.

```python
# --- VARIABLES ---

# Asignamos variables con el signo '='
# No necesitamos declarar el tipo (int, float); Python lo infiere.

voltaje = 220       # Entero (int)
resistencia = 10.5  # Decimal (float)
sistema_activo = True # Booleano (lógica binaria: 1 o 0)
nombre_pozo = "Pozo Norte-X" # Texto (string)

# La función print() muestra el resultado en pantalla
print("El voltaje es:", voltaje)

# La función type() nos dice qué tipo de dato es
print("Tipo de dato de resistencia:", type(resistencia))
```

**0.2 Operaciones Matemáticas y Librerías:** Al igual que Matlab, Python funciona como una calculadora científica:

```python
# --- CÁLCULOS EN PYTHON ---

# Ley de Ohm: I = V / R
corriente = voltaje / resistencia

# Potencia: P = V * I
potencia = voltaje * corriente

print("La corriente calculada es", corriente, "Amperios")
print("La potencia es:", potencia, "Watts")
```

A diferencia de Matlab, la mayoría de funciones de Python están en librerías, y antes de usarlas, debemos importar esas librerías:

```python
# --- LIBRERÍAS ---

# Calculemos el voltaje a partir de la potencia y la resistencia.
# Para ello debemos usar la función "raíz cuadrada", sqrt().
# Esta función está disponible en la librería "math", así que primero debemos importar esta librería.
# Además, podemos darle nombre un "alias" personalizado a la librería, por ejemplo "matem"

import math as matem

# Voltaje calculado: voltaje_calculado = sqrt(potencia*resistencia)
voltaje_calculado = matem.sqrt(potencia*resistencia)

print(f"El voltaje calculado es: {voltaje_calculado:.2f} Voltios") #.2f se usa para mostrar 2 decimales
```

**0.3 Estructuras de Datos: Listas y Diccionarios:**
En ingeniería rara vez trabajamos con un solo dato; trabajamos con vectores o series. En Python, usamos listas (que definimos usando corchetes []) y diccionarios (definidas usando llaves {}).

Las listas son los "vectores" de Pyhton; es decir, colecciones ordenadas indexadas por posición (0,1,2...). Por otro lado, los diccionarios son colecciones desordenadas de pares "llave"-"valor".

**Nota:** En Python, los elementos de las listas se empiezan a contar desde CERO, no desde uno.

**0.4 Identación, Condicionales (if, else) y Ciclos (for):**
A diferencia de otros lenguajes que usan llaves {} para agrupar código, Python utiliza la indentación (sangría). Es obligatorio, no estético.

Cada vez que una línea termina en dos puntos (:), la siguiente línea debe tener identación (usualmente 4 espacios). Todo el código que comparte el mismo nivel de indentación pertenece al mismo bloque lógico.

```python
# --- CONDICIONALES ---

import random # importamos la librería random

retorno_diario = -random.randint(1, 10) # simulamos una caída aleatoria entre el 10% y 1%
print(f"retorno: {retorno_diario} %")

# Noten los dos puntos (:) al final de la línea 'if'
# Noten el espacio (4 espacios o 1 TAB) en las líneas print
if retorno_diario < -5:
    print("ALERTA: Caída significativa detectada.") # esta línea se ejecuta si es TRUE porque está identada respecto al "if"
    print("Revisar noticias macroeconómicas.") # esta línea también se ejecuta si es TRUE porque está identada
else:
    print("Mercado estable.") # esta línea se ejecuta si es FALSE porque está identada respecto al "else"

print("Fin del análisis.") # Esta línea se ejecuta SIEMPRE, porque no tiene indentación.
```

```python
# --- CICLOS ---

# Caso1: Usemos un ciclo for para operar sobre una lista
precios_USD = [100, 102, 98, 105, 110] # lista de precios en dolares
precios_COP = []

# "Para cada 'precio' dentro de la lista 'precios'..."
for precio in precios_USD:
    # Lo convertimos a pesos (multiplicamos por 3650)
    precio_COP = precio*3650
    precios_COP.append(precio_COP)

print("Lista de precios en pesos es", precios_COP)
```

```python
# Caso 2: Si no tienes una lista, pero quieres repetir algo 5 veces, usas la función range()
for numero in range(5): # range(5) crea una secuencia: 0, 1, 2, 3, 4
    print(f"Vuelta número: {numero}")
```

```python
# Caso 3: Usemos ahora un ciclo for para operar sobre un diccionario
ventas = {
    "Lunes": 100,
    "Martes": 250,
    "Miércoles": 180
}

# Si queremos recorrer el diccionario sobre las LLAVES:
for dia in ventas.keys():
  print(f"El día {dia} se registraron ventas")

# Si queremos recorrer el diccionario sobre los VALORES:
total_ventas = 0
for valor in ventas.values():
  total_ventas = total_ventas + valor
print(f"El total de ventas de la semana fue {total_ventas}")
```

**Nota:** Aunque el ciclo for es fundamental para entender la lógica, en Ciencia de Datos intentamos evitarlo cuando procesamos millones de datos. Preferimos la "vectorización" (operar sobre toda la columna a la vez). Esto lo veremos en el Módulo 1 (Pandas).

# **MÓDULO 1: PANDAS**

Las listas y diccionarios de Python son útiles, pero para trabajar con datos a un nivel más avanzado necesitamos algo más eficiente. Aquí entra Pandas, una librería de Python diseñada para trabajar con datos.

Pandas introduce el DataFrame: Es una tabla con etiquetas (filas y columnas), donde los tipos de datos están optimizados para manejar millones de registros. Es como una hoja de Excel, pero eficiente y programable. Además, en un DataFrame se pueden almacenar datos de diferentes tipos: números, fechas, strings, etc.


**1.1 Preliminares:**
Primero, importamos la librería y descargamos los datos con los que trabajaremos: el comportamiento del S&P 500, un indicador del estado de la economía mundial.  

```python
# --- IMPORTACIÓN DE PANDAS Y CARGA DE DATOS ---

import pandas as pd # Importamos la librería con el alias 'pd' (estándar mundial)
import yfinance as yf # Librería Yahoo Finance, para descargar datos financieros reales

# Descargamos datos históricos del índice S&P 500 (desde 1990 hasta 2025)
print("Descargando dataset...")
datos_crudos = yf.download("^GSPC", start="1990-01-01", end="2025-12-31")

# Veamos cómo son los datos que acabamos de descargar:
print(datos_crudos.head()) # .head muestra las primeras 5 filas de los datos descargados

# Exportemos a formato .CSV los datos de cierre de mercado 'Close'para simular la lectura de un archivo local
datos_crudos['Close'].to_csv('sp500_history.csv') # .to_csv() crea el archivo sp500_history.csv
print("Archivo 'sp500_history.csv' generado exitosamente.")
```

**1.2 Carga e Inspección:**
Ahora utilizaremos la función `read_csv` de la librería Pandas para cargar el archivo en un DataFrame.

```python
# --- CREACIÓN DEL DataFrame ---

# Creamos el DataFrame llamado "mercado" con la función de PANDAS .read_csv()
mercado = pd.read_csv('sp500_history.csv')
print("DataFrame 'mercado' generado exitosamente.")
```

Las columnas y filas de los **DataFrames** tienen etiquetas que nos ayudan a identificarlas:

* **Etiquetas de columna:** Generalmente son los nombres de las variables almacenadas (ej. `Edad`, `Precio`).
* **Etiquetas de fila (Índice):** Actúan como el identificador único de cada fila. Comúnmente son números consecutivos, pero también pueden ser fechas (en series temporales) o textos específicos.

Una característica clave es que las columnas pueden almacenar **diferentes tipos de datos**. Por ejemplo, la primera columna puede contener números enteros y la segunda cadenas de texto (*strings*).

```python
# --- EXPLORACIÓN DE CARACTERÍSTICAS DEL DataFrame ---

print(f"Dimensiones: {mercado.shape}") # .shape devuelve (número de filas, número de columnas)
display(mercado.head()) # .head muestra las primeras 5 filas

print("\n--- INFORMACIÓN GENERAL ---")
print(mercado.info()) # .info() muestra tipos de datos y memoria

print("\n--- NOMBRES DE COLUMNAS ---")
print(mercado.columns) # Muestra los nombres de las variables

print("\n--- ÍNDICE (ETIQUETAS DE FILAS) ---")
print(mercado.index) # Muestra cómo están etiquetadas las filas

# Acceder a una fila específica por su POSICIÓN (usamos .iloc)
print("\n--- DATO EN LA FILA 100 ---")
fila_100 = mercado.iloc[99] # Recordar: Python cuenta desde 0, así que la fila 100 es el índice 99
print(fila_100)

# Si solo queremos ver la fecha de esa fila:
print(f"\nFecha de la fila 100: {mercado.iloc[99]['Date']}")
```

**1.3 Manipulación del DataFrame:** Los datos rara vez vienen limpios o en el formato que queremos. Aquí simularemos un proceso de pre-procesamiento del DataFrame

```python
# --- CAMBIO DE ETIQUETAS DE LAS COLUMNAS DEL DataFrame ---

# Vamos a renombrar las columnas de nuestro DataFrame "mercado". La primera la etiquetaremos como "fecha" y la segunda como "precio"
mercado.rename(columns={'Date': 'fecha', '^GSPC': 'precio'}, inplace=True) # El parámetro inplace=True hace que el cambio se guarde en el mismo DataFrame.
display(mercado.head()) # .head muestra las primeras 5 filas

# Alternativa a inplace=True:
mercado = mercado.rename(columns={'Date': 'fecha', '^GSPC': 'precio'})
```

```python
# --- CAMBIO DE ÍNDICES (FILAS) DEL DataFrame ---
# Como nuestros datos son una serie temporal, es conveniente etiquetar las filas con la fecha en lugar de usar un número

# Convertimos la columna a formato fecha (importante para que Pandas entienda que son fechas y no texto)
mercado['fecha'] = pd.to_datetime(mercado['fecha'])

# Establecemos la columna 'fecha' como el índice
mercado_ind = mercado.set_index('fecha', drop=False) # Usamos drop=False para mantener la columna "fecha" (aunque es redundante...¿Por qué?)
display(mercado_ind.head()) # .head muestra las primeras 5 filas
```

```python
# --- CONSULTA POR ETIQUETA ---

# Definimos una fecha de interés (debe tener el formato 'AAAA-MM-DD')
fecha_interes = '2020-03-16'

# Usamos .loc[fila, columna] para buscar por ETIQUETA
# Fila = fecha_interes
# Columna = 'precio'
valor = mercado_ind.loc[fecha_interes, 'precio']

print(f"El precio del S&P 500 el día {fecha_interes} fue: {valor}")
```

**2.1 Creación de Variables (Feature Engineering):** Para analizar una serie de tiempo (financiera, demográfica o climática), el valor absoluto importa menos que la variación porcentual.
$$
  \text{variacion}=\frac{\text{precio_actual}-\text{precio_ayer}}{\text{precio_ayer}}
$$


```python
# --- VECTORIZACIÓN ---

# Pandas aplica la operación a las 9,000 filas instantáneamente.
# .shift(1) mueve los datos una posición hacia abajo (el precio de ayer).

mercado_ind['precio_ayer'] = mercado_ind['precio'].shift(1)

# Calculamos variación porcentual de forma manual (para entender la matemática)
mercado_ind['variacion'] = (mercado_ind['precio'] - mercado_ind['precio_ayer']) / mercado_ind['precio_ayer']

# Forma optimizada de Pandas (hace lo mismo que arriba)
mercado_ind['variacion'] = mercado_ind['precio'].pct_change()

# Limpiamos los datos nulos generados por el shift (la primera fila no tiene "ayer")
mercado_ind.dropna(inplace=True)

mercado_ind.head()
```

**2.2 Filtrado con Lógica Booleana: ** En lugar de un `if` dentro de un `for`, usamos filtros vectorizados.

Vamos a ver cuáles han sido los "Cisnes Negros" (eventos extremos negativos). Es decir, eventos donde el mercado ha caído más del 5% en un día

```python
# --- FILTRADO ---
# Definimos un umbral: caída mayor al 5%
umbral_crisis = -0.05

# Filtro: Esto devuelve un DataFrame sólo con las filas que cumplen la condición
crisis = mercado_ind[mercado_ind['variacion'] < umbral_crisis]

print(f"Se encontraron {len(crisis)} eventos extremos en el periodo.")
# Mostramos los 10 peores Cisnes Negros
print(crisis.sort_values(by='variacion', ascending=True).head(10))
```
